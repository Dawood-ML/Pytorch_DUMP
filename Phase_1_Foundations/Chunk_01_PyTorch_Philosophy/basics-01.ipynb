## We will start this notebook with tensorOPS with irirs datset:


scalars (single numbers), vectors (1D arrays), and matrices (2D arrays). These are the fundamental building blocks of all deep learning computations.

In PyTorch, the fundamental data structure that encompasses all of these is the `torch.Tensor`.

* A scalar is a 0-dimensional tensor.
* A vector is a 1-dimensional tensor.
* A matrix is a 2-dimensional tensor.
* A stack of matrices (like an RGB image) is a 3-dimensional tensor.

A batch of images is a 4-dimensional tensor.
Think of a torch.Tensor as a super-powered NumPy array. It can live on a GPU for massive acceleration and seamlessly works with PyTorch's automatic differentiation engine, which we'll cover soon.

Today, we'll convert the Iris dataset, which is stored in NumPy arrays, into these powerful PyTorch Tensors.

```python
import torch
import sklearn
from sklearn.datasets import load_iris
import numpy as np


print(f"Pytorch version : { torch.__version__ }")
print(f"sklearn - version : {sklearn.__version__}")



################################################
# Load the dataset

iris = load_iris()
X = iris.data
y = iris.target

print(f"X.datatype : {X.dtype}, y.datatype : {y.dtype}")


############################################
# Convertour dataset to pytorch tensors




# X is our feature matrix
# Neural networks typically expect input data to be float

X_tensor = torch.tensor(X, dtype=torch.float32)

# y is our label vector
# Labels for classification are often long integers
# Hence, the dtype 'long'
y_tensor = torch.tensor(y, dtype=torch.long)


# Let's see what we've created.
print("--- Tensor Properties ---")
print(f"Shape of features tensor: {X_tensor.shape}")
print(f"Shape of labels tensor: {y_tensor.shape}\n")

print(f"Data type of features tensor: {X_tensor.dtype}")
print(f"Data type of labels tensor: {y_tensor.dtype}\n")

# .device tells us if the tensor is on the CPU or a GPU. For now, it's CPU.
print(f"Device tensor is stored on: {X_tensor.device}\n")





# --- 4. See the Data ---
# Let's look at the first 5 rows of our data, now as a PyTorch Tensor.
print("--- First 5 Samples ---")
print("Features (X):")
print(X_tensor[:5])
print("\nLabels (y):")
print(y_tensor[:5])
print(f"Which correspond to flower types: {[iris.target_names[i] for i in y_tensor[:5]]}")
```

#### 6. Professional Best Practices Introduced
* Data Types Matter: We explicitly set torch.float32 for features and torch.long for labels. This is standard practice and prevents common errors later on. float32 is the default for model weights and is a good balance of precision and memory usage.

* Know Your Data's Home: The .device attribute is critical. All computations between tensors must happen on the same device (e.g., all on the CPU or all on a specific GPU). We'll master this later.

* Reproducibility (A Teaser): For truly reproducible results, you should set a random seed at the beginning of your script. Add torch.manual_seed(42) to the top of the script. We'll see why this is crucial when we start training models.
