```python
import torch
import sklearn
from sklearn.datasets import load_iris
import numpy as np


print(f"Pytorch version : { torch.__version__ }")
print(f"sklearn - version : {sklearn.__version__}")



################################################
# Load the dataset

iris = load_iris()
X = iris.data
y = iris.target

print(f"X.datatype : {X.dtype}, y.datatype : {y.dtype}")


############################################
# Convertour dataset to pytorch tensors




# X is our feature matrix
# Neural networks typically expect input data to be float

X_tensor = torch.tensor(X, dtype=torch.float64)

# y is our label vector
# Labels for classification are often long integers
# Hence, the dtype 'long'
y_tensor = torch.tensor(y, dtype=torch.long)


# Let's see what we've created.
print("--- Tensor Properties ---")
print(f"Shape of features tensor: {X_tensor.shape}")
print(f"Shape of labels tensor: {y_tensor.shape}\n")

print(f"Data type of features tensor: {X_tensor.dtype}")
print(f"Data type of labels tensor: {y_tensor.dtype}\n")

# .device tells us if the tensor is on the CPU or a GPU. For now, it's CPU.
print(f"Device tensor is stored on: {X_tensor.device}\n")





# --- 4. See the Data ---
# Let's look at the first 5 rows of our data, now as a PyTorch Tensor.
print("--- First 5 Samples ---")
print("Features (X):")
print(X_tensor[:5])
print("\nLabels (y):")
print(y_tensor[:5])
print(f"Which correspond to flower types: {[iris.target_names[i] for i in y_tensor[:5]]}")
```
